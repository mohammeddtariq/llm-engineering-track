To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
from openai import OpenAI
import requests
from IPython.display import Markdown, display, update_display
```

```python
# constants

OLLAMA_BASE_URL = "http://localhost:11434/v1"
MODEL_LLAMA = 'llama3.2'
```

```python
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
return sum(w * self.weights[i] for i, w in enumerate(inputs))
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
```

```python
# Get Llama 3.2 to answer, with streaming

system_prompt = "You are a helpful technical tutor who explains code and concepts clearly, with examples where useful."

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": question}
]

stream = ollama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=messages,
    stream=True
)

response = ""
handle = display(Markdown(""), display_id=True)
for chunk in stream:
    content = chunk.choices[0].delta.content or ""
    response += content
    update_display(Markdown(response), display_id=handle.display_id)
```
